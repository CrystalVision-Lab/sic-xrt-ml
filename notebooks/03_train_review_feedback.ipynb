# 프로그램 검수 자료 일괄 학습

검수는 자료를 쌓는 단계입니다. Data Tools에서 변환한 새 자료 폴더와 기존 개발 자료를 지정하세요. 웨이퍼 8은 최종 시험용이므로 입력하지 않습니다.


```python
from pathlib import Path

import torch

from sic_xrt_ml.training.human_feedback import inspect, run

DATASET = Path("검수자료폴더")
RETAINED = Path("기존개발자료폴더")
VALIDATION_WAFER = "2"
EPOCHS = 15
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
OUTPUT = Path("outputs/review_training_new")
RUN_TRAINING = False
```

```python
if (DATASET / "dataset.json").exists() and (RETAINED / "cohort.json").exists():
    readiness = inspect(DATASET, RETAINED, VALIDATION_WAFER)[-1]
    print(readiness)
else:
    readiness = {"can_train": False}
    print("먼저 검수 자료·기존 개발 자료의 경로를 지정하세요.")
```

학습 가능 여부를 확인한 다음 RUN_TRAINING을 True로 바꾸어 아래 셀을 실행하세요. 부족한 확인 항목은 별도로 표시됩니다. 최소 수량은 기능 시험 조건이며 충분한 품질을 보장하지 않습니다.


```python
if RUN_TRAINING and readiness["can_train"]:
    report = run(DATASET, RETAINED, VALIDATION_WAFER, OUTPUT, EPOCHS, device=DEVICE)
    print(report["status"])
    print(report["human_validation"])
    print("현재 모델은 자동 교체되지 않습니다. 보고서:", OUTPUT / "report.json")
else:
    print("학습은 실행하지 않았습니다. 자료 확인 후 RUN_TRAINING=True로 설정하세요.")
```

이 결과는 사람이 확인한 후보 패치에 대한 평가입니다. 전체 영상의 누락률·독립 정확도와 구분하세요. 종류 미확정은 종류 학습에서 제외하고 위치 미확정은 위치 손실에서 제외합니다. 기존 자료도 함께 학습하여 이전 자료를 보존합니다.
